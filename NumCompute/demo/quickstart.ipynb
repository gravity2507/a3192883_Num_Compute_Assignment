# NumCompute quickstart


```python
from numcompute.io import read_csv
from numcompute.preprocessing import Imputer, StandardScaler, OneHotEncoder
from numcompute.sort_search import topk, quickselect, binary_search
from numcompute.rank import rank, percentile
from numcompute.stats import describe, histogram, quantiles
from numcompute.optim import grad, jacobian
import numpy as np
X = read_csv('data/stream_data.csv')[:,:3]
X = Imputer().fit_transform(X)
X = StandardScaler().fit_transform(X)
print('Top k:',topk(X[:,0],3,return_indices=False))
print('Quickselect:',quickselect(X[:,0],3))
print('Binary search:',binary_search(np.sort(X[:,0]),0))
print('Ranks:',rank([2,2,3]),'Percentile:',percentile(X[:,0],50))
print('Stats:',describe(X)['mean'],'Histogram:',histogram(X[:,0])[0])
print('Quantiles:',quantiles(X[:,0],[.25,.5,.75]))
print('One-hot:',OneHotEncoder().fit_transform([['red'],['blue']]))
print('Gradient:',grad(lambda z:np.sum(z*z),np.array([1.,2.])))
print('Jacobian:',jacobian(lambda z:2*z,np.array([1.,2.])))
```
