# Streaming decision tree demo



```python
from numcompute.io import iter_csv_chunks
from numcompute.preprocessing import Imputer
from numcompute.tree import DecisionTreeClassifier
from numcompute.pipeline import Pipeline
from numcompute.stream import StreamTrainer
from numcompute.visualise import plot_metric_over_time
model = Pipeline([('impute',Imputer()),('tree',DecisionTreeClassifier(max_depth=4,buffer_size=100))])
trainer = StreamTrainer(model)
for chunk in iter_csv_chunks('data/stream_data.csv',chunk_size=50):
    trainer.fit_chunk(chunk[:,:-1],chunk[:,-1].astype(int))
print(trainer.metrics.result())
plot_metric_over_time([r['cumulative_accuracy'] for r in trainer.history])
```
